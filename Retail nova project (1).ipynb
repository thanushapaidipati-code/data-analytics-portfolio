```python

# Step 1: Import libraries
import pandas as pd
import numpy as np

# Step 2: Load dataset
df = pd.read_csv("retail_nova.csv")

# Display original data
print("Original Shape:", df.shape)
print("\nMissing Values:")
print(df.isnull().sum())

# Step 3: Replace ?? with NaN in Unit_Price
df["Unit_Price"] = df["Unit_Price"].replace("??", np.nan)
df["Unit_Price"] = pd.to_numeric(df["Unit_Price"], errors="coerce")

# Step 4: Handle missing values
for col in ["City", "Channel", "Product_Category"]:
    df[col] = df[col].fillna(df[col].mode()[0])

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Unit_Price"] = df["Unit_Price"].fillna(df["Unit_Price"].median())
df["Notes"] = df["Notes"].fillna("No Notes")

# Step 5: Standardize Return_Flag
df["Return_Flag"] = (
    df["Return_Flag"].astype("string").str.strip().str.lower()
)

df["Return_Flag"] = df["Return_Flag"].replace({
    "y": "Yes",
    "yes": "Yes",
    "n": "No",
    "no": "No"
})

df["Return_Flag"] = df["Return_Flag"].fillna("Unknown")
df.loc[~df["Return_Flag"].isin(["Yes", "No"]), "Return_Flag"] = "Unknown"

# Step 6: Convert dates
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")
df["Delivery_Date"] = pd.to_datetime(df["Delivery_Date"], errors="coerce")

# Step 7: Calculate Total Sales
df["Total_Sales"] = (
    df["Quantity"] * df["Unit_Price"] * (1 - df["Discount"])
)

# Step 8: Calculate Delivery Days
df["Delivery_Days"] = (
    df["Delivery_Date"] - df["Order_Date"]
).dt.days

# Step 9: Handle negative delivery days
print("Negative Delivery Days:", (df["Delivery_Days"] < 0).sum())

df.loc[df["Delivery_Days"] < 0, "Delivery_Days"] = np.nan

# Step 10: Check duplicates
print("Duplicate Rows:", df.duplicated().sum())

# Step 11: Check data types
print("\nData Types:")
print(df.dtypes)

# Step 12: Final missing values
print("\nFinal Missing Values:")
print(df.isnull().sum())

# Step 13: Final shape
print("\nFinal Shape:", df.shape)

# Step 14: Export cleaned data
df.to_excel("cleaned_sales_data_AI.xlsx", index=False)

print("\nCleaning completed successfully!")
```
